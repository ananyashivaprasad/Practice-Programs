```python
def my_function(country="Norway"):
    print("I am from"+country)
my_function("Sweden")
my_function("India")
my_function()
my_function("Brazil")
```

```python
#Sample code 1 for keyword arguments
def student(fname, Iname):
    print(fname, Iname)
student(fname='Good', Iname='Practice')
student(fname='Practice', Iname='Good')
```

```python
#Sample code for positional arguments
def nameAge(name, age):
    print("Hi I am", name)
    print("My age", age)
print("Case=1")
nameAge("Suraj", 27)
print("\nCase-2")
nameAge(27, "Suraj")
```

```python
#Sample code for arbitrary keyword arugument
def myFun(*argv):
    for arg in argv:
        print(arg)
myFun('Hello', 'welcome', 'to', 'MIT')
```

```python
def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n-1)
print(factorial(4))
```
