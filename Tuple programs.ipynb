```python
#Program to print the 4th element from first and 4th element from last in a tuple
numbers=[10,20,20,40,50,60,70,80,90]
print("Tuple:", numbers)
print("4th element from the first:",numbers[3])
print("4th element from the last:", numbers[-4])
```

```python
#To check whether an element exists in a tuple
list_num=[10,20,30,40,50,60,70,80]
num=int(input("Enter a number from the list:"))
if num in list_num:
    print("The number requested:")
else:
    print("Number requested does not exist")
```

```python
#Program to convert list into a tuple
elements1=[10,20,30,40]
elements_tuple=tuple(elements1)
print("Given list is:", elements1)
print("Given tuple is:", elements_tuple)
```

```python
#Program to find index of an item in a tuple
numbers_tuple=(10,20,30,40,50)
num=int(input("Enter a number from the list"))
if num in numbers_tuple:
    x=numbers_tuple.index(num)
    print("The index of the number:",x)
else:
 print("Invalid")
```

```python
#Program to find the maximum and minimum element in a tuple
numbers=(25,20,45,5,60)
maximum=max(numbers)
minimum=min(numbers)
print("Tuple",numbers)
print("Maximum element:", maximum)
print("Minimum element:",minimum)
```

```python
#program to swap the first and last element of a tuple
numbers=(10,20,30,40,50,60)
new_tuple=(numbers[-1],)+ numbers[1:-1]+(numbers[0],)
print("Original Tuple:",numbers)
print("Tuple after swapping:", new_tuple)
```
